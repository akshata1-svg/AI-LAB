```python
def get_neighbors(state):

    neighbors = []
    zero_idx = state.index(0)
    row, col = divmod(zero_idx, 3)


    moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]

    for dr, dc in moves:
        new_row, new_col = row + dr, col + dc
        if 0 <= new_row < 3 and 0 <= new_col < 3:
            new_idx = new_row * 3 + new_col
            new_state = list(state)
            new_state[zero_idx], new_state[new_idx] = new_state[new_idx], new_state[zero_idx]
            neighbors.append(tuple(new_state))

    return neighbors

def depth_limited_search(state, goal_state, limit, path, visited):

    if state == goal_state:
        return path

    if limit <= 0:
        return None

    for neighbor in get_neighbors(state):

        if neighbor not in visited:
            visited.add(neighbor)
            result = depth_limited_search(neighbor, goal_state, limit - 1, path + [neighbor], visited)
            if result is not None:
                return result
            visited.remove(neighbor)

    return None

def ids_8_puzzle(start_state, goal_state, max_possible_depth=50):

    start_tuple = tuple(start_state)
    goal_tuple = tuple(goal_state)


    for depth in range(max_possible_depth):
        visited = {start_tuple}
        result = depth_limited_search(start_tuple, goal_tuple, depth, [start_tuple], visited)

        if result is not None:
            return result

    return None

def print_board(state):

    for i in range(0, 9, 3):
        print(f"{state[i]} {state[i+1]} {state[i+2]}")
    print("-" * 10)


if __name__ == "__main__":

    initial_state = (1, 2, 3,
                     0, 4, 6,
                     7, 5, 8)

    target_goal   = (1, 2, 3,
                     4, 5, 6,
                     7, 8, 0)

    print("Searching for the optimal solution using Iterative Deepening Search (IDS)...")
    solution_path = ids_8_puzzle(initial_state, target_goal)

    if solution_path:
        print(f"Success! Optimal solution found in {len(solution_path) - 1} moves.\n")
        for step, state in enumerate(solution_path):
            print(f"Step {step}:")
            print_board(state)
    else:
        print("No solution found within the maximum depth limit.")
```
