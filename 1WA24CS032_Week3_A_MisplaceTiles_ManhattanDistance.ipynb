```python

import heapq

goal = (1, 2, 3, 4, 5, 6, 7, 8, 0)

def h1(s):
    return sum(s[i] != 0 and s[i] != goal[i] for i in range(9))

def h2(s):
    d = 0
    for i, v in enumerate(s):
        if v:
            x, y = divmod(i, 3)
            a, b = divmod(v - 1, 3)
            d += abs(x - a) + abs(y - b)
    return d

def neighbors(s):
    z = s.index(0)
    r, c = divmod(z, 3)
    result = []

    for dr, dc in [(0, 1), (0, -1), (1, 0), (-1, 0)]:
        nr, nc = r + dr, c + dc
        if 0 <= nr < 3 and 0 <= nc < 3:
            j = nr * 3 + nc
            t = list(s)
            t[z], t[j] = t[j], t[z]
            result.append(tuple(t))

    return result

def astar(start, heuristic):
    pq = [(heuristic(start), 0, start, [])]
    best = {start: 0}

    while pq:
        f, g, state, path = heapq.heappop(pq)

        if state in path:
            continue

        path = path + [state]

        if state == goal:
            return path

        for nxt in neighbors(state):
            ng = g + 1

            if ng < best.get(nxt, float('inf')):
                best[nxt] = ng
                heapq.heappush(
                    pq, (ng + heuristic(nxt), ng, nxt, path)
                )

    return None

start = (1, 2, 3, 4, 0, 6, 7, 5, 8)

for name, h in [("Misplaced Tiles", h1),
                ("Manhattan Distance", h2)]:
    path = astar(start, h)

    print("\n", name)
    if path:
        print("Moves:", len(path) - 1)
        for s in path:
            for i in range(0, 9, 3):
                print(*s[i:i+3])
            print()
    else:
        print("No solution")
```
