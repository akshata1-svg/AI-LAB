```python

from itertools import product


def implies(a, b):
    return (not a) or b

def knowledge_base(P, Q, R):
    return (
        implies(P, Q)
        and implies(Q, R)
        and P
    )

def query(P, Q, R):
    return R

def truth_table_entailment():
    symbols = ['P', 'Q', 'R']
    entailment = True

    print("P Q R | P->Q | Q->R | KB | Query")
    print("----------------------------------")


    for values in product([True, False], repeat=3):

        model = dict(zip(symbols, values))

        P = model['P']
        Q = model['Q']
        R = model['R']


        kb_value = knowledge_base(P, Q, R)
        query_value = query(P, Q, R)


        print(
            int(P), int(Q), int(R), "|",
            int(implies(P, Q)), "  |",
            int(implies(Q, R)), "  |",
            int(kb_value), " |",
            int(query_value)
        )


        if kb_value and not query_value:
            entailment = False

    print("\nResult:")

    if entailment:
        print("KB entails the query (KB |= R)")
    else:
        print("KB does NOT entail the query")

truth_table_entailment()
```
