```python
board = [' '] * 9
player = 'X'

def show():
    print(f" {board[0]} | {board[1]} | {board[2]} ")
    print("-----------")
    print(f" {board[3]} | {board[4]} | {board[5]} ")
    print("-----------")
    print(f" {board[6]} | {board[7]} | {board[8]} ")

wins = [
    [0,1,2], [3,4,5], [6,7,8],
    [0,3,6], [1,4,7], [2,5,8],
    [0,4,8], [2,4,6]
]

for turn in range(9):
    show()
    while True:
        try:
            pos = int(input(f"Player {player}, choose 0-8: "))
            if 0 <= pos <= 8 and board[pos] == ' ':
                board[pos] = player
                break
            else:
                print("Taken or invalid. Try again.")
        except ValueError:
            print("Number please.")


    for w in wins:
        if board[w[0]] == board[w[1]] == board[w[2]] == player:
            show()
            print(f"Player {player} wins!")
            exit()

    player = 'O' if player == 'X' else 'X'

show()
print("It's a draw!")
```
